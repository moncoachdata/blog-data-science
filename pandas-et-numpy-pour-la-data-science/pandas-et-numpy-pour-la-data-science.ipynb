## NumPy

### 1. argpartition()

```python
# Tableau arbitraire
x = np.array([12, 10, 12, 0, 6, 8, 9, 1, 16, 4, 6, 0])

index_val = np.argpartition(x, -4)[-4:]
print(index_val)
```

```python
# Les valeurs correspondantes, triées
print(np.sort(x[index_val]))
```

### 2. allclose()

Le troisième argument est la tolérance relative `rtol`.

```python
array1 = np.array([0.12, 0.17, 0.24, 0.29])
array2 = np.array([0.13, 0.19, 0.26, 0.31])

# avec une tolérance relative de 0,1 : False
print(np.allclose(array1, array2, rtol=0.1))

# avec une tolérance relative de 0,2 : True
print(np.allclose(array1, array2, rtol=0.2))
```

```python
np.isclose(array1, array2, rtol=0.1)
```

### 3. clip()

```python
x = np.array([3, 17, 14, 23, 2, 2, 6, 8, 1, 2, 16, 0])

np.clip(x, 2, 5)
```

### 4. extract()

```python
# Nombres entiers aléatoires (avec une graine pour obtenir toujours les mêmes)
rng = np.random.default_rng(42)
arr = rng.integers(20, size=12)
print(arr)

# Diviser par 2 et vérifier si le reste est égal à 1 (nombres impairs)
cond = np.mod(arr, 2) == 1
print(cond)

# Utiliser extract() pour obtenir les valeurs
print(np.extract(cond, arr))

# Appliquer directement une condition (| signifie « ou »)
print(np.extract((arr < 3) | (arr > 15), arr))
```

```python
# extract() revient à filtrer avec un masque booléen
print(np.array_equal(np.extract(cond, arr), arr[cond]))
```

### 5. where()

```python
y = np.array([1, 5, 6, 8, 1, 7, 3, 6, 9])

# Positions des valeurs supérieures à 5
np.where(y > 5)
```

```python
# Avec trois arguments : une valeur si la condition est vraie, une autre sinon
np.where(y > 5, "OK", "NON")
```

### 6. percentile()

```python
a = np.array([1, 5, 6, 8, 1, 7, 3, 6, 9])
print("50e percentile de a :", np.percentile(a, 50))

b = np.array([[10, 7, 4], [3, 2, 1]])
print("30e percentile de b, axis=0 :", np.percentile(b, 30, axis=0))
print("30e percentile de b, axis=1 :", np.percentile(b, 30, axis=1))
```

```python
print(np.quantile(a, 0.5))
```

## pandas

### 7. read_csv(nrows=)

```python
# Jeu de données en ligne : nombre mensuel de passagers aériens de 1949 à 1960
url = "https://raw.githubusercontent.com/moncoachdata/blog-data-science/main/pandas-et-numpy-pour-la-data-science/AirPassengers.csv"

# On ne lit que les 10 premières lignes
df = pd.read_csv(url, nrows=10, index_col=0)
df
```

```python
# On peut aussi ne garder que certaines colonnes
pd.read_csv(url, usecols=["value"], nrows=3)
```

### 8. map()

```python
# Un DataFrame de nombres aléatoires (avec une graine)
rng = np.random.default_rng(0)
dframe = pd.DataFrame(rng.standard_normal((4, 3)).round(2), columns=list("bde"),
                      index=["Inde", "USA", "Chine", "Russie"])
dframe
```

```python
# Une fonction qui formate un nombre avec 2 décimales
changefn = lambda x: "%.2f" % x

# Changement valeur par valeur sur la colonne d
dframe["d"].map(changefn)
```

```python
# map() accepte aussi un dictionnaire
continents = {"Inde": "Asie", "USA": "Amérique", "Chine": "Asie", "Russie": "Europe"}
dframe.index.map(continents)
```

```python
# Sur un DataFrame entier, on utilise DataFrame.map() (anciennement applymap())
dframe.map(changefn)
```

### 9. apply()

```python
# Écart entre le maximum et le minimum
fn = lambda x: x.max() - x.min()

# Appliqué à chaque colonne du DataFrame
dframe.apply(fn)
```

```python
# Appliqué à chaque ligne avec axis=1
dframe.apply(fn, axis=1)
```

### 10. isin()

```python
# On réutilise le DataFrame lu avec read_csv
df[df["value"].isin([112, 118, 132])]
```

```python
# ~ inverse le filtre : toutes les lignes sauf celles-ci
df[~df["value"].isin([112, 118, 132])]
```

### 11. copy()

Depuis pandas 3.0, la copie sur écriture est active par défaut.

```python
# Créer une série d'exemple
data = pd.Series(["Inde", "Pakistan", "Chine", "Mongolie"])

# Une simple affectation ne copie rien : data1 et data sont le même objet
data1 = data
print(data1 is data)

# Changer une valeur via data1 change aussi data
data1[0] = "USA"
print(data)
```

```python
# Pour obtenir un objet indépendant, on utilise copy()
new = data.copy()

# attribution d'une nouvelle valeur dans la copie
new[1] = "Valeur changée"

print(new)
print(data)
```

```python
# Avec pandas 3, un sous-ensemble est lui aussi indépendant (copie sur écriture)
sous = data[:2]
sous[0] = "Autre"
print(sous)
print(data)
```

## Bonus : pivot_table()

```python
# Un DataFrame d'exemple : prénom, diplôme, âge, ville
school = pd.DataFrame({
    "Prénom": ["Jay", "Usher", "Nicky", "Romero", "Will"],
    "Diplôme": ["Master", "Licence", "Licence", "Master", "Licence"],
    "Âge": [26, 22, 20, 23, 24],
    "Ville": ["Zurich", "Genève", "Zurich", "Zurich", "Genève"],
})
school
```

```python
# Âge moyen par diplôme
pd.pivot_table(school, values="Âge", index="Diplôme", aggfunc="mean")
```

```python
# Diplômes en lignes, villes en colonnes, âge moyen dans les cellules
table = pd.pivot_table(school, values="Âge", index="Diplôme", columns="Ville",
                       aggfunc="mean", fill_value=0)
table
```

```python
# Plusieurs agrégations, avec les totaux (margins)
pd.pivot_table(school, values="Âge", index="Diplôme", aggfunc=["mean", "count"], margins=True)
```
