## 1. Comment créer une liste en une seule ligne avec une compréhension de liste ?

Écrire une boucle `for` chaque fois que tu as besoin de définir une liste est fastidieux. Heureusement, Python permet de le faire en une seule ligne de code.

La syntaxe peut paraître difficile au début, mais une fois familiarisé avec cette technique, tu l'utiliseras très souvent.

Regarde l'exemple ci-dessous, qui présente les deux syntaxes : la création d'une liste en plusieurs lignes, puis la compréhension de liste en une seule ligne.

Dans cet exemple, on souhaite mettre au carré les nombres impairs d'une liste. Soit `numbers` une liste contenant quelques nombres :

````python
numbers = [1, 2, 3, 4, 5]

squared_odds = []

for n in numbers:
    if n % 2 == 1:
        squared_odds.append(n ** 2)

print(squared_odds)
````

Ce code fait plusieurs choses :

- on assigne une liste vide à la variable `squared_odds` ;
- on y ajoute les valeurs correspondantes avec la méthode `append` ;
- on exécute une boucle `for` pour parcourir les éléments de la liste `numbers` ;
- on vérifie avec une condition `if` que le nombre est impair.

Avec une compréhension de liste, ce code devient :

````python
numbers = [1, 2, 3, 4, 5]

squared_odds = [n ** 2 for n in numbers if n % 2 == 1]

print(squared_odds)
````

## 2. Qu'est-ce qu'une fonction lambda ?

En Python, une fonction lambda est une fonction « anonyme » : elle n'est pas définie par un nom.

Alors que les fonctions « normales » sont définies avec le mot-clé `def` suivi du nom de la fonction, les fonctions anonymes sont définies avec le mot-clé `lambda`.

### À quoi elles servent ?

Les fonctions `lambda` servent à créer de petites fonctions ponctuelles et anonymes. Elles te permettent d'appliquer une fonction sans avoir à la définir.

La syntaxe de base des fonctions `lambda` est :

```python
lambda arguments: expression
```

Une fonction `lambda` peut faire tout ce que fait une fonction normale, à condition de tenir en une seule expression.

Voici un exemple simple :

````python
double = lambda x: x * 2
print(double(5))
````

## 3. Comment utiliser map et filter avec une fonction lambda ?

Une fois que tu maîtrises les fonctions `lambda`, les associer aux fonctions `map` et `filter` est très puissant.

### Que fait la fonction map ?

La fonction `map` prend une liste et la transforme en une nouvelle liste, en effectuant une opération sur chaque élément (à partir d'une fonction `lambda`).

Dans l'exemple ci-dessous, `map` parcourt chaque élément d'une liste et retourne une nouvelle liste dans laquelle chaque élément est multiplié par 2. Note que la fonction `list` convertit le résultat en objet de type liste.

````python
# map
seq = [1, 2, 3, 4, 5]

result = list(map(lambda var: var*2, seq))
print(result)
````

### Que fait la fonction filter ?

La fonction `filter` prend une liste et applique une règle (une condition). Comme `map`, sauf qu'elle renvoie un sous-ensemble de la liste d'origine : les éléments pour lesquels la règle est vraie.

Dans l'exemple ci-dessous, la règle est une fonction qui ne garde que les valeurs supérieures à 2. On obtient donc la sous-liste des valeurs plus grandes que 2.

````python
# filtrer
seq = [1, 2, 3, 4, 5]

result = list(filter(lambda x: x > 2, seq))
print(result)
````

## 4. Quelle différence entre arange et linspace ?

Pour créer des tableaux NumPy rapidement, ne cherche pas plus loin que les fonctions `arange` et `linspace`. Chacune a un but précis, mais on les préfère à la fonction `range` car elles produisent des tableaux NumPy, plus faciles à utiliser en Data Science.

### Que fait arange ?

`arange` renvoie des valeurs espacées uniformément dans un intervalle donné. En plus d'un point de départ (`start`) et d'arrêt (`stop`), tu peux définir une taille de pas (`step`) ou un type de données.

Note que le point d'arrêt est une valeur de « coupure » : il n'est pas inclus dans le tableau résultant.

````python
import numpy as np

# np.arange(start, stop, step)
np.arange(3, 7, 2)
````

### Que fait linspace ?

`linspace` est très similaire, avec une légère différence : il renvoie des nombres espacés uniformément sur un intervalle spécifié. Étant donné un point de départ (`start`), un point d'arrêt (`stop`) et un nombre de valeurs (`num`), il les espace équitablement et renvoie un tableau NumPy contenant exactement `num` valeurs.

Cette fois-ci, la valeur `stop` est présente dans le tableau résultant. C'est particulièrement utile pour les visualisations de données et les axes de graphiques.

````python
import numpy as np

# np.linspace(start, stop, num)
np.linspace(2.0, 3.0, num=5)
````

## 5. Que signifient axis=0 et axis=1 ?

Tu as peut-être déjà rencontré ce problème en supprimant une colonne d'un DataFrame Pandas ou en additionnant des valeurs dans une matrice NumPy. Sinon, tu y seras confronté un jour.

Prenons l'exemple de la suppression d'une ligne ou d'une colonne :

````python
import pandas as pd

df = pd.DataFrame({'Column A': [1, 2, 3], 'Column B': [4, 5, 6]}, index=['Row A', 'Row B', 'Row C'])

# supprimer une ligne (axis=0)
print(df.drop('Row A', axis=0))

# supprimer une colonne (axis=1)
print(df.drop('Column A', axis=1))
````

Combien de fois as-tu écrit cette ligne sans savoir ce qu'elle signifiait ? Comme tu peux le déduire, on définit `axis=1` pour traiter les colonnes et `axis=0` pour les lignes. Mais pourquoi ?

Voici une astuce pour s'en souvenir : regarde l'attribut `shape`.

````python
df.shape
````

L'attribut `shape` d'un DataFrame renvoie un tuple : le nombre de lignes en première position (indice 0), le nombre de colonnes en deuxième (indice 1). C'est exactement la logique de l'axe : `axis=0` pour les lignes, `axis=1` pour les colonnes.

## 6. Quelle différence entre concat, merge et join ?

Si tu es familier avec SQL, ces concepts te seront plus faciles. Ces fonctions sont des moyens de combiner des données de manière spécifique, et il n'est pas toujours évident de savoir laquelle utiliser et quand.

### Concat

`concat` permet d'ajouter un ou plusieurs DataFrames les uns aux autres, en dessous ou à côté (selon la valeur de `axis`).

````python
import pandas as pd

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

# en dessous (axis=0, valeur par défaut)
print(pd.concat([df1, df2], ignore_index=True))

# à côté (axis=1)
print(pd.concat([df1, df2], axis=1))
````

### Merge

`merge` combine plusieurs DataFrames sur une ou plusieurs colonnes communes, qui servent de clé de jointure.

````python
clients = pd.DataFrame({'id': [1, 2, 3], 'nom': ['Alice', 'Bob', 'Chloé']})
achats = pd.DataFrame({'id': [1, 2, 2, 4], 'montant': [30, 15, 25, 40]})

# jointure sur la colonne commune 'id' (how='inner' par défaut)
print(pd.merge(clients, achats, on='id'))

# how='left' conserve tous les clients
print(pd.merge(clients, achats, on='id', how='left'))
````

### Join

`join`, tout comme `merge`, combine deux DataFrames. Mais il les joint en fonction de leurs index, plutôt que d'une colonne spécifiée.

````python
gauche = pd.DataFrame({'ventes': [10, 20, 30]}, index=['lundi', 'mardi', 'mercredi'])
droite = pd.DataFrame({'visiteurs': [100, 150, 120]}, index=['lundi', 'mardi', 'jeudi'])

# jointure sur l'index (how='left' par défaut)
print(gauche.join(droite))
````

Pour d'autres syntaxes, des exemples plus concrets et des cas particuliers, consulte l'excellente [documentation Pandas](https://pandas.pydata.org/docs/user_guide/merging.html).

## 7. À quoi sert apply dans Pandas ?

`apply` ressemble à `map`, mais adaptée aux DataFrames Pandas, ou plus précisément aux objets Series (assez similaires aux tableaux NumPy).

`apply` applique une fonction à chaque élément d'une colonne ou d'une ligne, selon le paramètre `axis`. C'est très utile pour formater et manipuler les valeurs d'une colonne entière, sans écrire de boucle.

````python
notes = pd.DataFrame({'maths': [12, 15, 9], 'physique': [14, 11, 16]}, index=['Alice', 'Bob', 'Chloé'])

# sur une colonne : on applique une fonction à chaque élément
print(notes['maths'].apply(lambda x: x * 5))

# sur le DataFrame : axis=0 par colonne, axis=1 par ligne
print(notes.apply(max, axis=0))
print(notes.apply(max, axis=1))
````

## 8. Comment créer un tableau croisé dynamique avec pivot_table ?

Dernier point, mais non des moindres : les tableaux croisés dynamiques. Si tu connais Microsoft Excel, tu en as sûrement déjà entendu parler.

La fonction Pandas `pivot_table` crée un tableau croisé dynamique, de type feuille de calcul, sous forme de DataFrame. Les niveaux du tableau sont stockés dans des objets `MultiIndex` sur l'index et les colonnes du DataFrame résultant.

````python
ventes = pd.DataFrame({
    'ville': ['Paris', 'Paris', 'Lyon', 'Lyon', 'Paris', 'Lyon'],
    'produit': ['A', 'B', 'A', 'B', 'A', 'A'],
    'montant': [100, 150, 80, 120, 60, 90],
})

# lignes : ville, colonnes : produit, valeurs : somme des montants
pd.pivot_table(ventes, index='ville', columns='produit', values='montant', aggfunc='sum')
````
